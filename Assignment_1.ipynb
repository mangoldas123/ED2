### Q3

```python
import numpy as np
import matplotlib.pyplot as plt


c=1 
tau0=1 
v=np.linspace(0,0.9999,500)
gamma=1/np.sqrt(1-(v/c)**2)
tau_lab=gamma*tau0

plt.plot(v,tau_lab,label="Observed Lifetime")
plt.axhline(tau0,ls="--",label="Proper lifetime")

for vel in [0.01, 0.5, 0.9999]:
    gamma_val=1/np.sqrt(1-vel**2)
    tau_val=gamma_val*tau0
    plt.plot(vel,tau_val,'ko')
    plt.text(vel+0.02,tau_val,f"v={vel}c\nτ={tau_val:.2f} τ₀")

plt.xlabel("Velocity v/c")
plt.ylabel("Observed Lifetime τ/τ₀")
plt.title("Particle Lifetime in a frame of velocity v")
plt.legend()
plt.grid(True)
plt.show()
```

## Q2

```python
import numpy as np
import matplotlib.pyplot as plt

v1=np.array([0.01,0.5,0.99])
for v in v1:
    phi=np.arctan(v)  
    x=np.array([1,0])
    t=np.array([0,1])  
    R= np.array([[np.cos(phi),-np.sin(phi)],[np.sin(phi),np.cos(phi)]])
    
    x_prime=np.dot(R,x)
    t_prime=np.dot(R,t)
    
    plt.plot([0,x_prime[0]],[0,x_prime[1]],'--',label="x'")
    plt.plot([0,t_prime[0]],[0,t_prime[1]],'--',label="t'")
    plt.axhline(0,color='black')
    plt.axvline(0,color='black')
    plt.xlabel("x-------->")
    plt.ylabel("t-------->")
    plt.legend()
    plt.title(f"Space time diagram for v={v}")
    plt.show()

```
